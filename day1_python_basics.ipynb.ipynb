```python
# Day 1 - Hour 1: Variables and data types
# Think of variables as CTE aliases in SQL

# Numbers
revenue = 100000
tax_rate = 0.18
net = revenue * (1 - tax_rate)
print(f"Net revenue: {net}")

# String
name = "Sam"
city = "Chennai"
print(f"{name} lives in {city}")

# Boolean
is_active = True
print(f"Active user: {is_active}")

# List = ordered collection (like a column of values)
scores = [88, 92, 71, 95, 60]
print(f"Scores: {scores}")
print(f"First score: {scores[0]}")
print(f"Last score: {scores[-1]}")

# Dict = one row of data (like a single SQL record)
user = {"id": 1, "name": "Sam", "city": "Chennai", "salary": 45000}
print(f"User: {user}")
print(f"Name: {user['name']}")
```

```python
# Day 1 - Hour 1: Lists and Dicts deeper
# List of dicts = a SQL table

employees = [
    {"id": 1, "name": "Sam",   "city": "Chennai", "salary": 45000},
    {"id": 2, "name": "Priya", "city": "Mumbai",  "salary": 72000},
    {"id": 3, "name": "Raj",   "city": "Chennai", "salary": 58000},
    {"id": 4, "name": "Anita", "city": "Delhi",   "salary": 91000},
    {"id": 5, "name": "Kiran", "city": "Chennai", "salary": 63000},
]

# Access like SQL columns
print(employees[0])           # first row
print(employees[0]["name"])   # first row, name column
print(employees[-1]["city"])  # last row, city column

# Loop through all rows — like SELECT * FROM employees
for emp in employees:
    print(f"{emp['name']} earns {emp['salary']}")
```

```python
# Day 1 - Hour 2: List Comprehensions
# This is SELECT ... FROM ... WHERE in Python

employees = [
    {"id": 1, "name": "Sam",   "city": "Chennai", "salary": 45000},
    {"id": 2, "name": "Priya", "city": "Mumbai",  "salary": 72000},
    {"id": 3, "name": "Raj",   "city": "Chennai", "salary": 58000},
    {"id": 4, "name": "Anita", "city": "Delhi",   "salary": 91000},
    {"id": 5, "name": "Kiran", "city": "Chennai", "salary": 63000},
]

# SELECT name FROM employees
names = [emp["name"] for emp in employees]
print("All names:", names)

# SELECT name FROM employees WHERE city = 'Chennai'
chennai = [emp["name"] for emp in employees if emp["city"] == "Chennai"]
print("Chennai employees:", chennai)

# SELECT name, salary FROM employees WHERE salary > 60000
high_earners = [
    {"name": emp["name"], "salary": emp["salary"]}
    for emp in employees
    if emp["salary"] > 60000
]
print("High earners:", high_earners)

# SELECT salary * 1.10 AS new_salary FROM employees (10% raise)
with_raise = [
    {"name": emp["name"], "new_salary": emp["salary"] * 1.10}
    for emp in employees
]
print("After raise:", with_raise)
```

```python
# Day 1 - Hour 2: Control flow
# if/elif/else = CASE WHEN in SQL

employees = [
    {"id": 1, "name": "Sam",   "city": "Chennai", "salary": 45000},
    {"id": 2, "name": "Priya", "city": "Mumbai",  "salary": 72000},
    {"id": 3, "name": "Raj",   "city": "Chennai", "salary": 58000},
    {"id": 4, "name": "Anita", "city": "Delhi",   "salary": 91000},
    {"id": 5, "name": "Kiran", "city": "Chennai", "salary": 63000},
]

# CASE WHEN salary > 80000 THEN 'Senior'
#      WHEN salary > 55000 THEN 'Mid'
#      ELSE 'Junior' END
def get_level(salary):
    if salary > 80000:
        return "Senior"
    elif salary > 55000:
        return "Mid"
    else:
        return "Junior"

# Apply to all employees
for emp in employees:
    level = get_level(emp["salary"])
    print(f"{emp['name']:10} | {emp['salary']:>8} | {level}")

# Now combine with list comprehension
seniors = [emp["name"] for emp in employees if get_level(emp["salary"]) == "Senior"]
print(f"\nSeniors: {seniors}")
```

```python
# Day 1 - Hour 3: Functions
# Functions you'll use in every ML pipeline

# Basic function with default argument
def calculate_tax(salary, rate=0.18):
    return salary * rate

print(calculate_tax(50000))         # uses default 0.18
print(calculate_tax(50000, 0.30))   # overrides default

# ---

# *args = accept any number of arguments
def total_revenue(*amounts):
    return sum(amounts)

print(total_revenue(1000, 2000, 3000))
print(total_revenue(500, 750))

# ---

# **kwargs = accept any number of keyword arguments
# sklearn uses this everywhere: model = LinearRegression(fit_intercept=True, normalize=False)
def create_profile(**details):
    for key, value in details.items():
        print(f"  {key}: {value}")

print("User profile:")
create_profile(name="Sam", city="Chennai", role="AI Engineer", salary=30000)

# ---

# Function that processes a list — the most common ML pattern
def clean_name(text, lowercase=True, strip=True):
    if strip:
        text = text.strip()
    if lowercase:
        text = text.lower()
    return text

raw_names = ["  SAM  ", "PRIYA ", " Raj", "ANITA  "]
cleaned = [clean_name(name) for name in raw_names]
print("\nCleaned names:", cleaned)
```

```python
# Day 1 - Hour 3: File I/O
# Reading CSV manually — so Pandas feels magical tomorrow

import csv

# First create a small CSV file
with open("employees.csv", "w") as f:
    f.write("name,city,salary\n")
    f.write("Sam,Chennai,45000\n")
    f.write("Priya,Mumbai,72000\n")
    f.write("Raj,Chennai,58000\n")
    f.write("Anita,Delhi,91000\n")
    f.write("Kiran,Chennai,63000\n")

print("CSV created successfully")

# Now read it back
with open("employees.csv") as f:
    rows = [row for row in csv.DictReader(f)]

print(f"Loaded {len(rows)} rows")
print("First row:", rows[0])

# Filter Chennai employees — same as before but from a file
chennai = [r for r in rows if r["city"] == "Chennai"]
print("Chennai employees:", [r["name"] for r in chennai])

# Calculate average salary
# Note: CSV values are strings — must convert to int first
salaries = [int(r["salary"]) for r in rows]
avg = sum(salaries) / len(salaries)
print(f"Average salary: {avg}")
```

```python
# Day 1 - Hour 4: OOP
# This is the exact pattern sklearn uses
# model = LinearRegression()  ← __init__
# model.fit(X, y)             ← method
# model.predict(X)            ← method

class DataPipeline:
    def __init__(self, filepath):
        self.filepath = filepath
        self.data = []
        self.row_count = 0
        print(f"Pipeline ready for: {filepath}")

    def load(self):
        import csv
        with open(self.filepath) as f:
            self.data = [row for row in csv.DictReader(f)]
        self.row_count = len(self.data)
        print(f"Loaded {self.row_count} rows")
        return self  # enables method chaining

    def filter(self, column, value):
        self.data = [row for row in self.data if row[column] == value]
        print(f"After filter: {len(self.data)} rows")
        return self

    def show(self):
        for row in self.data:
            print(row)
        return self

# Method chaining — exactly like pandas
# df.load().filter().show()
pipeline = DataPipeline("employees.csv")
pipeline.load().filter("city", "Chennai").show()
```
